```python
import os
import json
import yaml
import random
import warnings
import joblib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import (
    StandardScaler,
    LabelEncoder
)

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.cluster import KMeans

from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    silhouette_score,
    precision_score,
    recall_score,
    f1_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Dense,
    Dropout,
    Input
)

from tensorflow.keras.callbacks import EarlyStopping

from tensorflow.keras.utils import to_categorical


# =============================================================================
# 2. GLOBAL SETTINGS
# =============================================================================

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)


# =============================================================================
# 3. PROJECT DIRECTORY
# =============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\District-Level Electrical Motor Pump Adoption Pattern Analysis"
)

DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "5.49.csv"
)

os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# =============================================================================
# 4. PSO OUTPUT PATHS
# =============================================================================

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "pso_electrical_motor_pump_model.h5"
)

PKL_PATH = os.path.join(
    PROJECT_DIR,
    "pso_electrical_motor_pump_preprocessor.pkl"
)

YAML_PATH = os.path.join(
    PROJECT_DIR,
    "pso_electrical_motor_pump_config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR,
    "pso_electrical_motor_pump_metadata.json"
)


# CSV OUTPUTS

RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result.csv"
)

PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction.csv"
)

COMPARISON_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_model_comparison.csv"
)

FEATURE_SELECTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_feature_selection.csv"
)

FITNESS_HISTORY_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_fitness_history.csv"
)


# GRAPH OUTPUTS

ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_accuracy_graph.png"
)

LOSS_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_loss_graph.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_comparison_graph.png"
)

HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "pso_heatmap.png"
)

RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result_graph.png"
)

PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction_graph.png"
)

FEATURE_SELECTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_feature_selection_graph.png"
)

FITNESS_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_fitness_graph.png"
)

DISTRIBUTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_adoption_distribution_graph.png"
)

CONFUSION_MATRIX_PATH = os.path.join(
    PROJECT_DIR,
    "pso_confusion_matrix.png"
)


# =============================================================================
# 5. LOAD DATASET
# =============================================================================

print("=" * 80)
print("PSO-BASED DISTRICT-LEVEL ELECTRICAL MOTOR PUMP ADOPTION ANALYSIS")
print("=" * 80)

print("\nLoading dataset:")
print(DATASET_PATH)

df = pd.read_csv(
    DATASET_PATH
)

print("\nDataset loaded successfully.")
print("Dataset shape:", df.shape)

print("\nOriginal columns:")

for col in df.columns:
    print(" -", col)


# =============================================================================
# 6. CLEAN COLUMN NAMES
# =============================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

print("\nCleaned columns:")

for col in df.columns:
    print(" -", col)


# =============================================================================
# 7. REMOVE EMPTY COLUMNS
# =============================================================================

empty_columns = [
    col
    for col in df.columns
    if df[col].isna().all()
]

if len(empty_columns) > 0:

    print("\nRemoving empty columns:")
    print(empty_columns)

    df = df.drop(
        columns=empty_columns
    )


# =============================================================================
# 8. REMOVE DUPLICATES
# =============================================================================

original_rows = len(df)

df = (
    df
    .drop_duplicates()
    .reset_index(drop=True)
)

removed_duplicates = (
    original_rows - len(df)
)

print(
    "\nDuplicate rows removed:",
    removed_duplicates
)


# =============================================================================
# 9. CONVERT NUMERIC-LIKE COLUMNS
# =============================================================================

for col in df.columns:

    if df[col].dtype == "object":

        cleaned = (
            df[col]
            .astype(str)
            .str.strip()
            .str.replace(",", "", regex=False)
            .str.replace("%", "", regex=False)
        )

        converted = pd.to_numeric(
            cleaned,
            errors="coerce"
        )

        original_valid = (
            df[col]
            .notna()
            .sum()
        )

        numeric_valid = (
            converted
            .notna()
            .sum()
        )

        if (
            original_valid > 0
            and
            numeric_valid / original_valid >= 0.80
        ):
            df[col] = converted


# =============================================================================
# 10. IDENTIFY NUMERIC AND CATEGORICAL COLUMNS
# =============================================================================

numeric_columns = list(
    df.select_dtypes(
        include=np.number
    ).columns
)

categorical_columns = list(
    df.select_dtypes(
        exclude=np.number
    ).columns
)

print("\nNumeric columns:")

for col in numeric_columns:
    print(" -", col)

print("\nCategorical columns:")

for col in categorical_columns:
    print(" -", col)


# =============================================================================
# 11. HANDLE MISSING VALUES
# =============================================================================

for col in numeric_columns:

    median_value = (
        df[col].median()
    )

    if pd.isna(
        median_value
    ):
        median_value = 0

    df[col] = (
        df[col]
        .fillna(median_value)
    )


for col in categorical_columns:

    mode_values = (
        df[col].mode()
    )

    if len(mode_values) > 0:
        fill_value = (
            mode_values.iloc[0]
        )
    else:
        fill_value = "Unknown"

    df[col] = (
        df[col]
        .fillna(fill_value)
    )


print(
    "\nRemaining missing values:",
    int(
        df.isnull()
        .sum()
        .sum()
    )
)


# =============================================================================
# 12. IDENTIFY ELECTRICAL MOTOR PUMP TARGET
# =============================================================================

motor_columns = []

for col in df.columns:

    lower_col = (
        col.lower()
    )

    if (
        "motor" in lower_col
        and
        "pump" in lower_col
    ):
        motor_columns.append(
            col
        )


print("\nMotor pump related columns:")

for col in motor_columns:
    print(" -", col)


target_column = None


# Prefer percentage column

for col in motor_columns:

    lower_col = (
        col.lower()
    )

    if (
        "%" in col
        or
        "percent" in lower_col
    ):
        target_column = col
        break


if (
    target_column is None
    and
    len(motor_columns) > 0
):
    target_column = (
        motor_columns[-1]
    )


if target_column is None:

    raise ValueError(
        "Could not identify Electrical Motor Pump Use (%) column."
    )


print("\nSelected target indicator:")
print(target_column)


# =============================================================================
# 13. ENSURE TARGET IS NUMERIC
# =============================================================================

df[target_column] = pd.to_numeric(
    df[target_column],
    errors="coerce"
)

target_median = (
    df[target_column]
    .median()
)

if pd.isna(
    target_median
):
    target_median = 0

df[target_column] = (
    df[target_column]
    .fillna(target_median)
)


print("\nTarget statistics:")

print(
    df[target_column]
    .describe()
)


# =============================================================================
# 14. CORRELATION HEATMAP
# =============================================================================

print(
    "\nGenerating PSO correlation heatmap..."
)

heatmap_columns = list(
    df.select_dtypes(
        include=np.number
    ).columns
)

heatmap_columns = [
    col
    for col in heatmap_columns
    if df[col].nunique() > 1
]


if len(heatmap_columns) >= 2:

    correlation_matrix = (
        df[
            heatmap_columns
        ]
        .corr()
    )

    fig, ax = plt.subplots(
        figsize=(12, 9)
    )

    image = ax.imshow(
        correlation_matrix.values,
        aspect="auto",
        cmap="coolwarm",
        vmin=-1,
        vmax=1
    )

    ax.set_xticks(
        range(
            len(
                correlation_matrix.columns
            )
        )
    )

    ax.set_yticks(
        range(
            len(
                correlation_matrix.columns
            )
        )
    )

    ax.set_xticklabels(
        correlation_matrix.columns,
        rotation=45,
        ha="right",
        fontsize=8
    )

    ax.set_yticklabels(
        correlation_matrix.columns,
        fontsize=8
    )

    for i in range(
        len(
            correlation_matrix.columns
        )
    ):

        for j in range(
            len(
                correlation_matrix.columns
            )
        ):

            value = (
                correlation_matrix.iloc[
                    i,
                    j
                ]
            )

            ax.text(
                j,
                i,
                f"{value:.2f}",
                ha="center",
                va="center",
                fontsize=7
            )

    plt.colorbar(
        image,
        ax=ax,
        label="Correlation"
    )

    plt.title(
        "PSO - Correlation Heatmap",
        fontsize=14,
        fontweight="bold"
    )

    plt.tight_layout()

    plt.savefig(
        HEATMAP_PATH,
        dpi=300,
        bbox_inches="tight"
    )

    print(
        "Saved:",
        HEATMAP_PATH
    )

    plt.show()

    plt.close()

else:

    print(
        "Not enough numeric columns for heatmap."
    )


# =============================================================================
# 15. INITIAL NUMERIC FEATURES
# =============================================================================

numeric_columns = list(
    df.select_dtypes(
        include=np.number
    ).columns
)

initial_feature_columns = [
    col
    for col in numeric_columns
    if (
        col != target_column
        and
        df[col].nunique() > 1
    )
]


# =============================================================================
# 16. FALLBACK STATISTICAL FEATURES
# =============================================================================

if len(
    initial_feature_columns
) == 0:

    print(
        "\nNo independent varying numeric features found."
    )

    print(
        "Creating statistical pattern features..."
    )

    df[
        "Adoption_Rank"
    ] = (
        df[target_column]
        .rank(
            method="average"
        )
    )

    df[
        "Adoption_Deviation"
    ] = (
        df[target_column]
        -
        df[target_column].mean()
    )

    df[
        "Adoption_Absolute_Deviation"
    ] = np.abs(
        df[
            "Adoption_Deviation"
        ]
    )

    target_std = (
        df[target_column]
        .std()
    )

    if (
        pd.isna(target_std)
        or
        target_std == 0
    ):
        target_std = 1.0

    df[
        "Adoption_Z_Score"
    ] = (
        (
            df[target_column]
            -
            df[target_column].mean()
        )
        /
        target_std
    )

    initial_feature_columns = [
        "Adoption_Rank",
        "Adoption_Deviation",
        "Adoption_Absolute_Deviation",
        "Adoption_Z_Score"
    ]


print("\nCandidate features before PSO:")

for feature in initial_feature_columns:
    print(" -", feature)


# =============================================================================
# 17. CREATE ADOPTION CLASSES USING K-MEANS
# =============================================================================

print("\n" + "=" * 80)
print("ADOPTION PATTERN DISCOVERY")
print("=" * 80)

adoption_values = (
    df[
        [
            target_column
        ]
    ]
    .values
)

adoption_scaler = (
    StandardScaler()
)

adoption_scaled = (
    adoption_scaler
    .fit_transform(
        adoption_values
    )
)

unique_adoption_values = (
    df[target_column]
    .nunique()
)


if unique_adoption_values >= 3:
    n_clusters = 3

elif unique_adoption_values == 2:
    n_clusters = 2

else:
    n_clusters = 1


if n_clusters > 1:

    kmeans = KMeans(
        n_clusters=n_clusters,
        random_state=RANDOM_STATE,
        n_init=20
    )

    df[
        "Cluster"
    ] = (
        kmeans
        .fit_predict(
            adoption_scaled
        )
    )

else:

    df[
        "Cluster"
    ] = 0


# =============================================================================
# 18. CONVERT CLUSTERS TO ADOPTION LEVELS
# =============================================================================

cluster_means = (
    df
    .groupby(
        "Cluster"
    )[
        target_column
    ]
    .mean()
    .sort_values()
)

ordered_clusters = list(
    cluster_means.index
)

cluster_mapping = {}


if len(
    ordered_clusters
) == 1:

    cluster_mapping[
        ordered_clusters[0]
    ] = "Uniform Adoption"


elif len(
    ordered_clusters
) == 2:

    cluster_mapping[
        ordered_clusters[0]
    ] = "Low Adoption"

    cluster_mapping[
        ordered_clusters[1]
    ] = "High Adoption"


else:

    cluster_mapping[
        ordered_clusters[0]
    ] = "Low Adoption"

    cluster_mapping[
        ordered_clusters[1]
    ] = "Moderate Adoption"

    cluster_mapping[
        ordered_clusters[2]
    ] = "High Adoption"


df[
    "Adoption_Level"
] = (
    df[
        "Cluster"
    ]
    .map(
        cluster_mapping
    )
)


print("\nAdoption class distribution:")

print(
    df[
        "Adoption_Level"
    ]
    .value_counts()
)


# =============================================================================
# 19. SILHOUETTE SCORE
# =============================================================================

silhouette = None

if (
    n_clusters > 1
    and
    df[
        "Cluster"
    ].nunique() > 1
):

    try:

        silhouette = (
            silhouette_score(
                adoption_scaled,
                df[
                    "Cluster"
                ]
            )
        )

        print(
            "\nSilhouette Score:",
            round(
                silhouette,
                4
            )
        )

    except Exception:

        silhouette = None


# =============================================================================
# 20. ADOPTION DISTRIBUTION GRAPH
# =============================================================================

distribution = (
    df[
        "Adoption_Level"
    ]
    .value_counts()
)

plt.figure(
    figsize=(9, 6)
)

bars = plt.bar(
    distribution.index,
    distribution.values
)

plt.xlabel(
    "Adoption Level"
)

plt.ylabel(
    "Number of Districts"
)

plt.title(
    "PSO - Electrical Motor Pump Adoption Distribution",
    fontweight="bold"
)

for bar in bars:

    height = (
        bar.get_height()
    )

    plt.text(
        bar.get_x()
        +
        bar.get_width() / 2,
        height,
        str(
            int(
                height
            )
        ),
        ha="center",
        va="bottom"
    )

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    DISTRIBUTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

print(
    "Saved:",
    DISTRIBUTION_GRAPH_PATH
)

plt.show()

plt.close()


# =============================================================================
# 21. LABEL ENCODING
# =============================================================================

label_encoder = (
    LabelEncoder()
)

y = (
    label_encoder
    .fit_transform(
        df[
            "Adoption_Level"
        ]
    )
)

class_names = list(
    label_encoder.classes_
)

num_classes = len(
    class_names
)


print("\nClasses:")

for i, class_name in enumerate(
    class_names
):

    print(
        i,
        "->",
        class_name
    )


# =============================================================================
# 22. PREPARE INITIAL FEATURE MATRIX
# =============================================================================

X_initial = (
    df[
        initial_feature_columns
    ]
    .copy()
)

initial_scaler = (
    StandardScaler()
)

X_initial_scaled = (
    initial_scaler
    .fit_transform(
        X_initial
    )
)

number_of_features = (
    X_initial_scaled.shape[1]
)


# =============================================================================
# 23. PARTICLE SWARM OPTIMIZATION PARAMETERS
# =============================================================================

print("\n" + "=" * 80)
print("PARTICLE SWARM OPTIMIZATION FEATURE SELECTION")
print("=" * 80)


PSO_SWARM_SIZE = 30

PSO_ITERATIONS = 30

PSO_INERTIA_MAX = 0.9

PSO_INERTIA_MIN = 0.4

PSO_COGNITIVE = 2.0

PSO_SOCIAL = 2.0


print(
    "\nPSO Parameters:"
)

print(
    "Swarm Size:",
    PSO_SWARM_SIZE
)

print(
    "Iterations:",
    PSO_ITERATIONS
)

print(
    "Maximum Inertia:",
    PSO_INERTIA_MAX
)

print(
    "Minimum Inertia:",
    PSO_INERTIA_MIN
)

print(
    "Cognitive Coefficient:",
    PSO_COGNITIVE
)

print(
    "Social Coefficient:",
    PSO_SOCIAL
)


# =============================================================================
# 24. PSO FITNESS FUNCTION
# =============================================================================

def evaluate_particle(
    particle,
    X_data,
    y_data
):

    """
    Evaluate one binary PSO particle.

    A particle is a binary feature-selection vector.

    Example:

        [1, 0, 1, 1]

    means:

        Feature 1 = selected
        Feature 2 = removed
        Feature 3 = selected
        Feature 4 = selected

    Fitness combines:
        95% predictive accuracy
         5% feature reduction

    Higher fitness is better.
    """

    selected_indices = np.where(
        particle == 1
    )[0]

    # No feature selected = invalid solution

    if len(
        selected_indices
    ) == 0:

        return 0.0

    X_selected = X_data[
        :,
        selected_indices
    ]

    class_counts_local = (
        pd.Series(
            y_data
        )
        .value_counts()
    )

    # Only one class
    if len(
        class_counts_local
    ) < 2:

        return 1.0

    min_class_count = int(
        class_counts_local.min()
    )

    number_splits = min(
        3,
        min_class_count
    )

    classifier = (
        RandomForestClassifier(
            n_estimators=100,
            random_state=RANDOM_STATE,
            class_weight="balanced"
        )
    )

    if number_splits >= 2:

        try:

            cv = (
                StratifiedKFold(
                    n_splits=number_splits,
                    shuffle=True,
                    random_state=RANDOM_STATE
                )
            )

            scores = (
                cross_val_score(
                    classifier,
                    X_selected,
                    y_data,
                    cv=cv,
                    scoring="accuracy"
                )
            )

            predictive_accuracy = float(
                np.mean(
                    scores
                )
            )

        except Exception:

            classifier.fit(
                X_selected,
                y_data
            )

            predictions = (
                classifier.predict(
                    X_selected
                )
            )

            predictive_accuracy = (
                accuracy_score(
                    y_data,
                    predictions
                )
            )

    else:

        classifier.fit(
            X_selected,
            y_data
        )

        predictions = (
            classifier.predict(
                X_selected
            )
        )

        predictive_accuracy = (
            accuracy_score(
                y_data,
                predictions
            )
        )

    feature_ratio = (
        len(
            selected_indices
        )
        /
        number_of_features
    )

    feature_reduction = (
        1.0
        -
        feature_ratio
    )

    fitness = (
        0.95
        *
        predictive_accuracy
        +
        0.05
        *
        feature_reduction
    )

    return float(
        fitness
    )


# =============================================================================
# 25. SIGMOID FUNCTION FOR BINARY PSO
# =============================================================================

def sigmoid(
    value
):

    # Clip to avoid overflow

    value = np.clip(
        value,
        -20,
        20
    )

    return (
        1.0
        /
        (
            1.0
            +
            np.exp(
                -value
            )
        )
    )


# =============================================================================
# 26. INITIALIZE PSO SWARM
# =============================================================================

positions = np.random.randint(
    0,
    2,
    size=(
        PSO_SWARM_SIZE,
        number_of_features
    )
)


# Ensure every particle has at least one feature

for particle_index in range(
    PSO_SWARM_SIZE
):

    if np.sum(
        positions[
            particle_index
        ]
    ) == 0:

        random_feature = (
            np.random.randint(
                0,
                number_of_features
            )
        )

        positions[
            particle_index,
            random_feature
        ] = 1


# Initial velocities

velocities = np.random.uniform(
    low=-1.0,
    high=1.0,
    size=(
        PSO_SWARM_SIZE,
        number_of_features
    )
)


# =============================================================================
# 27. INITIAL PERSONAL BEST
# =============================================================================

personal_best_positions = (
    positions.copy()
)

personal_best_scores = np.zeros(
    PSO_SWARM_SIZE
)


for particle_index in range(
    PSO_SWARM_SIZE
):

    personal_best_scores[
        particle_index
    ] = evaluate_particle(
        positions[
            particle_index
        ],
        X_initial_scaled,
        y
    )


# =============================================================================
# 28. INITIAL GLOBAL BEST
# =============================================================================

best_particle_index = int(
    np.argmax(
        personal_best_scores
    )
)

global_best_position = (
    personal_best_positions[
        best_particle_index
    ]
    .copy()
)

global_best_score = float(
    personal_best_scores[
        best_particle_index
    ]
)


print(
    "\nInitial Global Best Fitness:",
    round(
        global_best_score,
        4
    )
)


# =============================================================================
# 29. RUN PARTICLE SWARM OPTIMIZATION
# =============================================================================

pso_history = []


for iteration in range(
    PSO_ITERATIONS
):

    # Linearly decrease inertia
    inertia_weight = (
        PSO_INERTIA_MAX
        -
        (
            (
                PSO_INERTIA_MAX
                -
                PSO_INERTIA_MIN
            )
            *
            iteration
            /
            max(
                1,
                PSO_ITERATIONS - 1
            )
        )
    )

    current_scores = []

    # -------------------------------------------------------------------------
    # EVALUATE CURRENT PARTICLES
    # -------------------------------------------------------------------------

    for particle_index in range(
        PSO_SWARM_SIZE
    ):

        current_position = (
            positions[
                particle_index
            ]
        )

        current_score = (
            evaluate_particle(
                current_position,
                X_initial_scaled,
                y
            )
        )

        current_scores.append(
            current_score
        )

        # Update personal best

        if (
            current_score
            >
            personal_best_scores[
                particle_index
            ]
        ):

            personal_best_scores[
                particle_index
            ] = current_score

            personal_best_positions[
                particle_index
            ] = (
                current_position.copy()
            )

        # Update global best

        if (
            current_score
            >
            global_best_score
        ):

            global_best_score = (
                current_score
            )

            global_best_position = (
                current_position.copy()
            )


    # -------------------------------------------------------------------------
    # SAVE ITERATION HISTORY
    # -------------------------------------------------------------------------

    average_fitness = float(
        np.mean(
            current_scores
        )
    )

    selected_feature_count = int(
        np.sum(
            global_best_position
        )
    )

    pso_history.append(
        {
            "Iteration":
                iteration + 1,

            "Global_Best_Fitness":
                global_best_score,

            "Average_Fitness":
                average_fitness,

            "Selected_Features":
                selected_feature_count,

            "Inertia_Weight":
                inertia_weight
        }
    )


    print(
        f"Iteration {iteration + 1:02d}/{PSO_ITERATIONS} | "
        f"Global Best Fitness: {global_best_score:.4f} | "
        f"Average Fitness: {average_fitness:.4f} | "
        f"Selected Features: {selected_feature_count} | "
        f"Inertia: {inertia_weight:.4f}"
    )


    # -------------------------------------------------------------------------
    # UPDATE PARTICLE VELOCITIES AND POSITIONS
    # -------------------------------------------------------------------------

    for particle_index in range(
        PSO_SWARM_SIZE
    ):

        r1 = np.random.random(
            number_of_features
        )

        r2 = np.random.random(
            number_of_features
        )

        cognitive_component = (
            PSO_COGNITIVE
            *
            r1
            *
            (
                personal_best_positions[
                    particle_index
                ]
                -
                positions[
                    particle_index
                ]
            )
        )

        social_component = (
            PSO_SOCIAL
            *
            r2
            *
            (
                global_best_position
                -
                positions[
                    particle_index
                ]
            )
        )

        velocities[
            particle_index
        ] = (
            inertia_weight
            *
            velocities[
                particle_index
            ]
            +
            cognitive_component
            +
            social_component
        )

        # Keep velocity bounded

        velocities[
            particle_index
        ] = np.clip(
            velocities[
                particle_index
            ],
            -6,
            6
        )

        probabilities = sigmoid(
            velocities[
                particle_index
            ]
        )

        random_values = np.random.random(
            number_of_features
        )

        positions[
            particle_index
        ] = (
            random_values
            <
            probabilities
        ).astype(int)

        # Ensure at least one feature remains

        if np.sum(
            positions[
                particle_index
            ]
        ) == 0:

            random_feature = (
                np.random.randint(
                    0,
                    number_of_features
                )
            )

            positions[
                particle_index,
                random_feature
            ] = 1


# =============================================================================
# 30. SAVE PSO FITNESS HISTORY
# =============================================================================

pso_history_df = (
    pd.DataFrame(
        pso_history
    )
)

pso_history_df.to_csv(
    FITNESS_HISTORY_CSV_PATH,
    index=False
)

print(
    "\nPSO fitness history saved:"
)

print(
    FITNESS_HISTORY_CSV_PATH
)


# =============================================================================
# 31. PSO FITNESS GRAPH
# =============================================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    pso_history_df[
        "Iteration"
    ],
    pso_history_df[
        "Global_Best_Fitness"
    ],
    marker="o",
    linewidth=2,
    label="Global Best Fitness"
)

plt.plot(
    pso_history_df[
        "Iteration"
    ],
    pso_history_df[
        "Average_Fitness"
    ],
    linewidth=2,
    label="Average Swarm Fitness"
)

plt.xlabel(
    "PSO Iteration"
)

plt.ylabel(
    "Fitness"
)

plt.title(
    "Particle Swarm Optimization Convergence",
    fontweight="bold"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    FITNESS_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

print(
    "Saved:",
    FITNESS_GRAPH_PATH
)

plt.show()

plt.close()


# =============================================================================
# 32. EXTRACT PSO-SELECTED FEATURES
# =============================================================================

best_particle = (
    global_best_position.copy()
)

best_fitness = (
    global_best_score
)

selected_indices = np.where(
    best_particle == 1
)[0]

selected_feature_columns = [
    initial_feature_columns[
        index
    ]
    for index in selected_indices
]


print("\n" + "=" * 80)
print("PSO FEATURE SELECTION RESULT")
print("=" * 80)

print(
    "\nBest PSO Fitness:",
    round(
        best_fitness,
        4
    )
)

print(
    "Features before PSO:",
    len(
        initial_feature_columns
    )
)

print(
    "Features after PSO:",
    len(
        selected_feature_columns
    )
)

print(
    "\nPSO Selected Features:"
)

for feature in selected_feature_columns:

    print(
        " ✓",
        feature
    )


# =============================================================================
# 33. FEATURE SELECTION CSV
# =============================================================================

feature_selection_data = []


for index, feature in enumerate(
    initial_feature_columns
):

    selected = int(
        best_particle[
            index
        ]
    )

    feature_selection_data.append(
        {
            "Feature":
                feature,

            "Selected":
                selected,

            "Status":
                (
                    "Selected"
                    if selected == 1
                    else "Removed"
                )
        }
    )


feature_selection_df = pd.DataFrame(
    feature_selection_data
)

feature_selection_df.to_csv(
    FEATURE_SELECTION_CSV_PATH,
    index=False
)

print(
    "\nPSO feature selection CSV saved:"
)

print(
    FEATURE_SELECTION_CSV_PATH
)


# =============================================================================
# 34. PSO FEATURE SELECTION GRAPH
# =============================================================================

plt.figure(
    figsize=(
        11,
        max(
            5,
            len(
                feature_selection_df
            )
            *
            0.6
        )
    )
)

bars = plt.barh(
    feature_selection_df[
        "Feature"
    ],
    feature_selection_df[
        "Selected"
    ]
)

plt.xlabel(
    "PSO Selection Status"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Particle Swarm Optimization Feature Selection",
    fontweight="bold"
)

plt.xticks(
    [
        0,
        1
    ],
    [
        "Removed",
        "Selected"
    ]
)

for bar, status in zip(
    bars,
    feature_selection_df[
        "Status"
    ]
):

    plt.text(
        bar.get_width()
        +
        0.02,
        bar.get_y()
        +
        bar.get_height() / 2,
        status,
        va="center"
    )

plt.xlim(
    0,
    1.3
)

plt.grid(
    axis="x",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    FEATURE_SELECTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

print(
    "Saved:",
    FEATURE_SELECTION_GRAPH_PATH
)

plt.show()

plt.close()


# =============================================================================
# 35. FINAL PSO FEATURE MATRIX
# =============================================================================

X = (
    df[
        selected_feature_columns
    ]
    .copy()
)

feature_scaler = (
    StandardScaler()
)

X_scaled = (
    feature_scaler
    .fit_transform(
        X
    )
)

print(
    "\nFinal PSO feature matrix shape:",
    X_scaled.shape
)


# =============================================================================
# 36. TRAIN / TEST SPLIT
# =============================================================================

class_counts = (
    pd.Series(
        y
    )
    .value_counts()
)

can_stratify = (
    len(
        class_counts
    ) > 1
    and
    class_counts.min() >= 2
)


if num_classes > 1:

    X_train, X_test, y_train, y_test = (
        train_test_split(
            X_scaled,
            y,
            test_size=0.25,
            random_state=RANDOM_STATE,
            stratify=(
                y
                if can_stratify
                else None
            )
        )
    )

else:

    X_train = X_scaled
    X_test = X_scaled

    y_train = y
    y_test = y


print(
    "\nTraining samples:",
    len(
        X_train
    )
)

print(
    "Testing samples:",
    len(
        X_test
    )
)


# =============================================================================
# 37. BUILD PSO-OPTIMIZED NEURAL NETWORK
# =============================================================================

if num_classes > 1:

    y_train_cat = (
        to_categorical(
            y_train,
            num_classes=num_classes
        )
    )

    y_test_cat = (
        to_categorical(
            y_test,
            num_classes=num_classes
        )
    )

    model = Sequential(
        [
            Input(
                shape=(
                    X_train.shape[1],
                )
            ),

            Dense(
                64,
                activation="relu"
            ),

            Dropout(
                0.20
            ),

            Dense(
                32,
                activation="relu"
            ),

            Dropout(
                0.15
            ),

            Dense(
                16,
                activation="relu"
            ),

            Dense(
                num_classes,
                activation="softmax"
            )
        ]
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.001
        ),
        loss="categorical_crossentropy",
        metrics=[
            "accuracy"
        ]
    )

    print(
        "\nPSO-Optimized Neural Network:"
    )

    model.summary()


# =============================================================================
# 38. TRAIN NEURAL NETWORK
# =============================================================================

history = None


if num_classes > 1:

    early_stopping = (
        EarlyStopping(
            monitor="val_loss",
            patience=30,
            restore_best_weights=True,
            verbose=1
        )
    )

    history = model.fit(
        X_train,
        y_train_cat,
        validation_split=0.20,
        epochs=250,
        batch_size=8,
        callbacks=[
            early_stopping
        ],
        verbose=1
    )


# =============================================================================
# 39. TEST PREDICTIONS
# =============================================================================

if num_classes > 1:

    test_probabilities = (
        model.predict(
            X_test,
            verbose=0
        )
    )

    test_predictions = (
        np.argmax(
            test_probabilities,
            axis=1
        )
    )

    nn_accuracy = (
        accuracy_score(
            y_test,
            test_predictions
        )
    )

    nn_precision = (
        precision_score(
            y_test,
            test_predictions,
            average="weighted",
            zero_division=0
        )
    )

    nn_recall = (
        recall_score(
            y_test,
            test_predictions,
            average="weighted",
            zero_division=0
        )
    )

    nn_f1 = (
        f1_score(
            y_test,
            test_predictions,
            average="weighted",
            zero_division=0
        )
    )

else:

    test_predictions = y_test

    nn_accuracy = 1.0
    nn_precision = 1.0
    nn_recall = 1.0
    nn_f1 = 1.0


print(
    "\nPSO Neural Network Accuracy:",
    f"{nn_accuracy * 100:.2f}%"
)

print(
    "Precision:",
    f"{nn_precision * 100:.2f}%"
)

print(
    "Recall:",
    f"{nn_recall * 100:.2f}%"
)

print(
    "F1 Score:",
    f"{nn_f1 * 100:.2f}%"
)


# =============================================================================
# 40. CLASSIFICATION REPORT
# =============================================================================

if num_classes > 1:

    print(
        "\nClassification Report:"
    )

    print(
        classification_report(
            y_test,
            test_predictions,
            labels=np.arange(
                num_classes
            ),
            target_names=class_names,
            zero_division=0
        )
    )


# =============================================================================
# 41. CONFUSION MATRIX
# =============================================================================

if num_classes > 1:

    cm = confusion_matrix(
        y_test,
        test_predictions,
        labels=np.arange(
            num_classes
        )
    )

    plt.figure(
        figsize=(8, 6)
    )

    plt.imshow(
        cm,
        cmap="Blues"
    )

    plt.title(
        "PSO Model Confusion Matrix",
        fontweight="bold"
    )

    plt.colorbar()

    plt.xticks(
        range(
            len(
                class_names
            )
        ),
        class_names,
        rotation=30,
        ha="right"
    )

    plt.yticks(
        range(
            len(
                class_names
            )
        ),
        class_names
    )

    plt.xlabel(
        "Predicted Class"
    )

    plt.ylabel(
        "Actual Class"
    )

    for i in range(
        cm.shape[0]
    ):

        for j in range(
            cm.shape[1]
        ):

            plt.text(
                j,
                i,
                str(
                    cm[
                        i,
                        j
                    ]
                ),
                ha="center",
                va="center"
            )

    plt.tight_layout()

    plt.savefig(
        CONFUSION_MATRIX_PATH,
        dpi=300,
        bbox_inches="tight"
    )

    print(
        "Saved:",
        CONFUSION_MATRIX_PATH
    )

    plt.show()

    plt.close()


# =============================================================================
# 42. ACCURACY GRAPH
# =============================================================================

if history is not None:

    plt.figure(
        figsize=(10, 6)
    )

    plt.plot(
        history.history[
            "accuracy"
        ],
        linewidth=2,
        label="Training Accuracy"
    )

    plt.plot(
        history.history[
            "val_accuracy"
        ],
        linewidth=2,
        label="Validation Accuracy"
    )

    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Accuracy"
    )

    plt.title(
        "PSO Neural Network Training and Validation Accuracy",
        fontweight="bold"
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        ACCURACY_GRAPH_PATH,
        dpi=300,
        bbox_inches="tight"
    )

    print(
        "Saved:",
        ACCURACY_GRAPH_PATH
    )

    plt.show()

    plt.close()


# =============================================================================
# 43. LOSS GRAPH
# =============================================================================

if history is not None:

    plt.figure(
        figsize=(10, 6)
    )

    plt.plot(
        history.history[
            "loss"
        ],
        linewidth=2,
        label="Training Loss"
    )

    plt.plot(
        history.history[
            "val_loss"
        ],
        linewidth=2,
        label="Validation Loss"
    )

    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Loss"
    )

    plt.title(
        "PSO Neural Network Training and Validation Loss",
        fontweight="bold"
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        LOSS_GRAPH_PATH,
        dpi=300,
        bbox_inches="tight"
    )

    print(
        "Saved:",
        LOSS_GRAPH_PATH
    )

    plt.show()

    plt.close()


# =============================================================================
# 44. MODEL COMPARISON
# =============================================================================

print("\n" + "=" * 80)
print("PSO MODEL COMPARISON")
print("=" * 80)

model_results = []


# -----------------------------------------------------------------------------
# PSO + NEURAL NETWORK
# -----------------------------------------------------------------------------

model_results.append(
    {
        "Model":
            "PSO + Neural Network",

        "Accuracy":
            nn_accuracy * 100,

        "Precision":
            nn_precision * 100,

        "Recall":
            nn_recall * 100,

        "F1_Score":
            nn_f1 * 100
    }
)


# =============================================================================
# 45. PSO + RANDOM FOREST
# =============================================================================

if num_classes > 1:

    rf_model = (
        RandomForestClassifier(
            n_estimators=200,
            random_state=RANDOM_STATE,
            class_weight="balanced"
        )
    )

    rf_model.fit(
        X_train,
        y_train
    )

    rf_predictions = (
        rf_model.predict(
            X_test
        )
    )

    rf_accuracy = (
        accuracy_score(
            y_test,
            rf_predictions
        )
    )

    rf_precision = (
        precision_score(
            y_test,
            rf_predictions,
            average="weighted",
            zero_division=0
        )
    )

    rf_recall = (
        recall_score(
            y_test,
            rf_predictions,
            average="weighted",
            zero_division=0
        )
    )

    rf_f1 = (
        f1_score(
            y_test,
            rf_predictions,
            average="weighted",
            zero_division=0
        )
    )

    model_results.append(
        {
            "Model":
                "PSO + Random Forest",

            "Accuracy":
                rf_accuracy * 100,

            "Precision":
                rf_precision * 100,

            "Recall":
                rf_recall * 100,

            "F1_Score":
                rf_f1 * 100
        }
    )


# =============================================================================
# 46. PSO + DECISION TREE
# =============================================================================

if num_classes > 1:

    dt_model = (
        DecisionTreeClassifier(
            random_state=RANDOM_STATE
        )
    )

    dt_model.fit(
        X_train,
        y_train
    )

    dt_predictions = (
        dt_model.predict(
            X_test
        )
    )

    dt_accuracy = (
        accuracy_score(
            y_test,
            dt_predictions
        )
    )

    dt_precision = (
        precision_score(
            y_test,
            dt_predictions,
            average="weighted",
            zero_division=0
        )
    )

    dt_recall = (
        recall_score(
            y_test,
            dt_predictions,
            average="weighted",
            zero_division=0
        )
    )

    dt_f1 = (
        f1_score(
            y_test,
            dt_predictions,
            average="weighted",
            zero_division=0
        )
    )

    model_results.append(
        {
            "Model":
                "PSO + Decision Tree",

            "Accuracy":
                dt_accuracy * 100,

            "Precision":
                dt_precision * 100,

            "Recall":
                dt_recall * 100,

            "F1_Score":
                dt_f1 * 100
        }
    )


# =============================================================================
# 47. PSO + KNN
# =============================================================================

if num_classes > 1:

    number_neighbors = min(
        5,
        len(
            X_train
        )
    )

    knn_model = (
        KNeighborsClassifier(
            n_neighbors=number_neighbors
        )
    )

    knn_model.fit(
        X_train,
        y_train
    )

    knn_predictions = (
        knn_model.predict(
            X_test
        )
    )

    knn_accuracy = (
        accuracy_score(
            y_test,
            knn_predictions
        )
    )

    knn_precision = (
        precision_score(
            y_test,
            knn_predictions,
            average="weighted",
            zero_division=0
        )
    )

    knn_recall = (
        recall_score(
            y_test,
            knn_predictions,
            average="weighted",
            zero_division=0
        )
    )

    knn_f1 = (
        f1_score(
            y_test,
            knn_predictions,
            average="weighted",
            zero_division=0
        )
    )

    model_results.append(
        {
            "Model":
                "PSO + KNN",

            "Accuracy":
                knn_accuracy * 100,

            "Precision":
                knn_precision * 100,

            "Recall":
                knn_recall * 100,

            "F1_Score":
                knn_f1 * 100
        }
    )


# =============================================================================
# 48. PSO + LOGISTIC REGRESSION
# =============================================================================

if num_classes > 1:

    lr_model = (
        LogisticRegression(
            max_iter=2000,
            random_state=RANDOM_STATE,
            class_weight="balanced"
        )
    )

    lr_model.fit(
        X_train,
        y_train
    )

    lr_predictions = (
        lr_model.predict(
            X_test
        )
    )

    lr_accuracy = (
        accuracy_score(
            y_test,
            lr_predictions
        )
    )

    lr_precision = (
        precision_score(
            y_test,
            lr_predictions,
            average="weighted",
            zero_division=0
        )
    )

    lr_recall = (
        recall_score(
            y_test,
            lr_predictions,
            average="weighted",
            zero_division=0
        )
    )

    lr_f1 = (
        f1_score(
            y_test,
            lr_predictions,
            average="weighted",
            zero_division=0
        )
    )

    model_results.append(
        {
            "Model":
                "PSO + Logistic Regression",

            "Accuracy":
                lr_accuracy * 100,

            "Precision":
                lr_precision * 100,

            "Recall":
                lr_recall * 100,

            "F1_Score":
                lr_f1 * 100
        }
    )


# =============================================================================
# 49. CREATE MODEL COMPARISON DATAFRAME
# =============================================================================

comparison_df = (
    pd.DataFrame(
        model_results
    )
    .sort_values(
        "Accuracy",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

comparison_df.to_csv(
    COMPARISON_CSV_PATH,
    index=False
)

print(
    "\nPSO Model Comparison:"
)

print(
    comparison_df
)

print(
    "\nSaved:",
    COMPARISON_CSV_PATH
)


# =============================================================================
# 50. MODEL COMPARISON GRAPH
# =============================================================================

plt.figure(
    figsize=(12, 7)
)

x = np.arange(
    len(
        comparison_df
    )
)

width = 0.20

plt.bar(
    x - 1.5 * width,
    comparison_df[
        "Accuracy"
    ],
    width,
    label="Accuracy"
)

plt.bar(
    x - 0.5 * width,
    comparison_df[
        "Precision"
    ],
    width,
    label="Precision"
)

plt.bar(
    x + 0.5 * width,
    comparison_df[
        "Recall"
    ],
    width,
    label="Recall"
)

plt.bar(
    x + 1.5 * width,
    comparison_df[
        "F1_Score"
    ],
    width,
    label="F1 Score"
)

plt.xlabel(
    "PSO-Based Model"
)

plt.ylabel(
    "Performance (%)"
)

plt.title(
    "PSO-Based Model Performance Comparison",
    fontweight="bold"
)

plt.xticks(
    x,
    comparison_df[
        "Model"
    ],
    rotation=25,
    ha="right"
)

plt.ylim(
    0,
    110
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

print(
    "Saved:",
    COMPARISON_GRAPH_PATH
)

plt.show()

plt.close()


# =============================================================================
# 51. FULL DATASET PREDICTIONS
# =============================================================================

if num_classes > 1:

    full_probabilities = (
        model.predict(
            X_scaled,
            verbose=0
        )
    )

    full_prediction_codes = (
        np.argmax(
            full_probabilities,
            axis=1
        )
    )

    full_prediction_labels = (
        label_encoder
        .inverse_transform(
            full_prediction_codes
        )
    )

    full_confidence = (
        np.max(
            full_probabilities,
            axis=1
        )
        *
        100
    )

else:

    full_prediction_codes = y

    full_prediction_labels = np.array(
        [
            class_names[0]
        ]
        *
        len(
            df
        )
    )

    full_confidence = (
        np.ones(
            len(
                df
            )
        )
        *
        100
    )


# =============================================================================
# 52. CREATE PSO RESULT CSV
# =============================================================================

result_df = (
    df.copy()
)

result_df[
    "PSO_Actual_Adoption_Level"
] = (
    df[
        "Adoption_Level"
    ]
)

result_df[
    "PSO_Predicted_Adoption_Level"
] = (
    full_prediction_labels
)

result_df[
    "PSO_Prediction_Confidence_Percent"
] = np.round(
    full_confidence,
    2
)

result_df[
    "PSO_Prediction_Correct"
] = (
    result_df[
        "PSO_Actual_Adoption_Level"
    ]
    ==
    result_df[
        "PSO_Predicted_Adoption_Level"
    ]
)

result_df.to_csv(
    RESULT_CSV_PATH,
    index=False
)

print(
    "\nPSO result CSV saved:"
)

print(
    RESULT_CSV_PATH
)

print(
    "\nResult preview:"
)

print(
    result_df.head()
)


# =============================================================================
# 53. RESULT GRAPH
# =============================================================================

actual_codes = (
    label_encoder
    .transform(
        result_df[
            "PSO_Actual_Adoption_Level"
        ]
    )
)

predicted_codes = (
    label_encoder
    .transform(
        result_df[
            "PSO_Predicted_Adoption_Level"
        ]
    )
)

record_numbers = np.arange(
    1,
    len(
        result_df
    )
    +
    1
)

plt.figure(
    figsize=(14, 7)
)

plt.plot(
    record_numbers,
    actual_codes,
    marker="o",
    linewidth=2,
    label="Actual Adoption Level"
)

plt.plot(
    record_numbers,
    predicted_codes,
    marker="x",
    linestyle="--",
    linewidth=2,
    label="PSO Predicted Adoption Level"
)

plt.yticks(
    range(
        len(
            class_names
        )
    ),
    class_names
)

plt.xlabel(
    "District Record"
)

plt.ylabel(
    "Adoption Level"
)

plt.title(
    "PSO - Actual vs Predicted Adoption Level",
    fontweight="bold"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

print(
    "Saved:",
    RESULT_GRAPH_PATH
)

plt.show()

plt.close()


# =============================================================================
# 54. CREATE PSO PREDICTION CSV
# =============================================================================

prediction_df = pd.DataFrame()


for col in categorical_columns:

    if col in df.columns:

        prediction_df[
            col
        ] = (
            df[
                col
            ]
        )


prediction_df[
    target_column
] = (
    df[
        target_column
    ]
)

prediction_df[
    "PSO_Predicted_Adoption_Level"
] = (
    full_prediction_labels
)

prediction_df[
    "PSO_Prediction_Confidence_Percent"
] = np.round(
    full_confidence,
    2
)

prediction_df.to_csv(
    PREDICTION_CSV_PATH,
    index=False
)

print(
    "\nPSO prediction CSV saved:"
)

print(
    PREDICTION_CSV_PATH
)

print(
    "\nPrediction preview:"
)

print(
    prediction_df.head()
)


# =============================================================================
# 55. PREDICTION GRAPH
# =============================================================================

predicted_numeric = (
    label_encoder
    .transform(
        prediction_df[
            "PSO_Predicted_Adoption_Level"
        ]
    )
)

plt.figure(
    figsize=(14, 7)
)

scatter = plt.scatter(
    record_numbers,
    prediction_df[
        target_column
    ],
    c=predicted_numeric,
    cmap="viridis",
    s=100,
    alpha=0.85
)

plt.plot(
    record_numbers,
    prediction_df[
        target_column
    ],
    alpha=0.4
)

plt.xlabel(
    "District Record"
)

plt.ylabel(
    target_column
)

plt.title(
    "PSO-Based Electrical Motor Pump Adoption Prediction",
    fontweight="bold"
)

colorbar = plt.colorbar(
    scatter
)

colorbar.set_label(
    "PSO Predicted Adoption Class"
)

colorbar.set_ticks(
    range(
        len(
            class_names
        )
    )
)

colorbar.set_ticklabels(
    class_names
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

print(
    "Saved:",
    PREDICTION_GRAPH_PATH
)

plt.show()

plt.close()


# =============================================================================
# 56. SAVE PSO H5 MODEL
# =============================================================================

print("\n" + "=" * 80)
print("SAVING PSO MODEL AND ARTIFACTS")
print("=" * 80)


if num_classes > 1:

    model.save(
        MODEL_PATH
    )

else:

    dummy_model = Sequential(
        [
            Input(
                shape=(
                    X_scaled.shape[1],
                )
            ),

            Dense(
                8,
                activation="relu"
            ),

            Dense(
                1,
                activation="sigmoid"
            )
        ]
    )

    dummy_model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=[
            "accuracy"
        ]
    )

    dummy_y = np.ones(
        len(
            X_scaled
        )
    )

    dummy_model.fit(
        X_scaled,
        dummy_y,
        epochs=10,
        batch_size=8,
        verbose=0
    )

    dummy_model.save(
        MODEL_PATH
    )


print(
    "\nPSO H5 model saved:"
)

print(
    MODEL_PATH
)


# =============================================================================
# 57. SAVE PSO PKL
# =============================================================================

preprocessor = {

    "method":
        "Binary Particle Swarm Optimization Feature Selection",

    "initial_scaler":
        initial_scaler,

    "feature_scaler":
        feature_scaler,

    "adoption_scaler":
        adoption_scaler,

    "label_encoder":
        label_encoder,

    "initial_feature_columns":
        initial_feature_columns,

    "selected_feature_columns":
        selected_feature_columns,

    "pso_best_particle":
        best_particle,

    "pso_best_fitness":
        best_fitness,

    "target_column":
        target_column,

    "cluster_mapping":
        cluster_mapping,

    "class_names":
        class_names,

    "random_state":
        RANDOM_STATE,

    "pso_parameters": {

        "swarm_size":
            PSO_SWARM_SIZE,

        "iterations":
            PSO_ITERATIONS,

        "inertia_max":
            PSO_INERTIA_MAX,

        "inertia_min":
            PSO_INERTIA_MIN,

        "cognitive_coefficient":
            PSO_COGNITIVE,

        "social_coefficient":
            PSO_SOCIAL
    }
}


if n_clusters > 1:

    preprocessor[
        "kmeans"
    ] = kmeans


joblib.dump(
    preprocessor,
    PKL_PATH
)

print(
    "\nPSO PKL saved:"
)

print(
    PKL_PATH
)


# =============================================================================
# 58. SAVE PSO YAML
# =============================================================================

yaml_data = {

    "project": {

        "name":
            "PSO-Based District-Level Electrical Motor Pump Adoption Pattern Analysis",

        "optimization":
            "Binary Particle Swarm Optimization",

        "random_state":
            int(
                RANDOM_STATE
            )
    },

    "dataset": {

        "path":
            DATASET_PATH,

        "rows":
            int(
                len(
                    df
                )
            ),

        "columns":
            int(
                len(
                    df.columns
                )
            ),

        "target":
            target_column
    },

    "pso": {

        "swarm_size":
            PSO_SWARM_SIZE,

        "iterations":
            PSO_ITERATIONS,

        "inertia_max":
            PSO_INERTIA_MAX,

        "inertia_min":
            PSO_INERTIA_MIN,

        "cognitive_coefficient":
            PSO_COGNITIVE,

        "social_coefficient":
            PSO_SOCIAL,

        "best_fitness":
            float(
                best_fitness
            ),

        "features_before":
            int(
                len(
                    initial_feature_columns
                )
            ),

        "features_after":
            int(
                len(
                    selected_feature_columns
                )
            ),

        "selected_features":
            selected_feature_columns
    },

    "model": {

        "type":
            "PSO Optimized Feedforward Neural Network",

        "framework":
            "TensorFlow Keras",

        "classes":
            class_names,

        "accuracy":
            float(
                nn_accuracy
            ),

        "precision":
            float(
                nn_precision
            ),

        "recall":
            float(
                nn_recall
            ),

        "f1_score":
            float(
                nn_f1
            )
    },

    "clustering": {

        "algorithm":
            "KMeans",

        "clusters":
            int(
                n_clusters
            ),

        "silhouette_score":
            (
                float(
                    silhouette
                )
                if silhouette is not None
                else None
            )
    },

    "outputs": {

        "model":
            MODEL_PATH,

        "preprocessor":
            PKL_PATH,

        "json":
            JSON_PATH,

        "result_csv":
            RESULT_CSV_PATH,

        "prediction_csv":
            PREDICTION_CSV_PATH,

        "comparison_csv":
            COMPARISON_CSV_PATH,

        "feature_selection_csv":
            FEATURE_SELECTION_CSV_PATH,

        "fitness_history_csv":
            FITNESS_HISTORY_CSV_PATH,

        "accuracy_graph":
            ACCURACY_GRAPH_PATH,

        "loss_graph":
            LOSS_GRAPH_PATH,

        "comparison_graph":
            COMPARISON_GRAPH_PATH,

        "heatmap":
            HEATMAP_PATH,

        "result_graph":
            RESULT_GRAPH_PATH,

        "prediction_graph":
            PREDICTION_GRAPH_PATH,

        "feature_selection_graph":
            FEATURE_SELECTION_GRAPH_PATH,

        "fitness_graph":
            FITNESS_GRAPH_PATH,

        "confusion_matrix":
            CONFUSION_MATRIX_PATH,

        "distribution_graph":
            DISTRIBUTION_GRAPH_PATH
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_data,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print(
    "\nPSO YAML saved:"
)

print(
    YAML_PATH
)


# =============================================================================
# 59. SAVE PSO JSON
# =============================================================================

json_data = {

    "project_name":
        "PSO-Based District-Level Electrical Motor Pump Adoption Pattern Analysis",

    "dataset":
        os.path.basename(
            DATASET_PATH
        ),

    "dataset_path":
        DATASET_PATH,

    "rows":
        int(
            len(
                df
            )
        ),

    "target_column":
        target_column,

    "optimization_algorithm":
        "Binary Particle Swarm Optimization",

    "pso_swarm_size":
        PSO_SWARM_SIZE,

    "pso_iterations":
        PSO_ITERATIONS,

    "pso_inertia_max":
        PSO_INERTIA_MAX,

    "pso_inertia_min":
        PSO_INERTIA_MIN,

    "pso_cognitive":
        PSO_COGNITIVE,

    "pso_social":
        PSO_SOCIAL,

    "pso_best_fitness":
        float(
            best_fitness
        ),

    "features_before_pso":
        int(
            len(
                initial_feature_columns
            )
        ),

    "features_after_pso":
        int(
            len(
                selected_feature_columns
            )
        ),

    "selected_features":
        selected_feature_columns,

    "classes":
        class_names,

    "number_of_clusters":
        int(
            n_clusters
        ),

    "silhouette_score":
        (
            float(
                silhouette
            )
            if silhouette is not None
            else None
        ),

    "neural_network_accuracy":
        float(
            nn_accuracy
        ),

    "precision":
        float(
            nn_precision
        ),

    "recall":
        float(
            nn_recall
        ),

    "f1_score":
        float(
            nn_f1
        ),

    "model_comparison":
        comparison_df.to_dict(
            orient="records"
        ),

    "generated_files": {

        "model":
            os.path.basename(
                MODEL_PATH
            ),

        "preprocessor":
            os.path.basename(
                PKL_PATH
            ),

        "yaml":
            os.path.basename(
                YAML_PATH
            ),

        "json":
            os.path.basename(
                JSON_PATH
            ),

        "result":
            os.path.basename(
                RESULT_CSV_PATH
            ),

        "prediction":
            os.path.basename(
                PREDICTION_CSV_PATH
            ),

        "comparison":
            os.path.basename(
                COMPARISON_CSV_PATH
            ),

        "feature_selection":
            os.path.basename(
                FEATURE_SELECTION_CSV_PATH
            ),

        "fitness_history":
            os.path.basename(
                FITNESS_HISTORY_CSV_PATH
            ),

        "accuracy_graph":
            os.path.basename(
                ACCURACY_GRAPH_PATH
            ),

        "loss_graph":
            os.path.basename(
                LOSS_GRAPH_PATH
            ),

        "comparison_graph":
            os.path.basename(
                COMPARISON_GRAPH_PATH
            ),

        "heatmap":
            os.path.basename(
                HEATMAP_PATH
            ),

        "result_graph":
            os.path.basename(
                RESULT_GRAPH_PATH
            ),

        "prediction_graph":
            os.path.basename(
                PREDICTION_GRAPH_PATH
            ),

        "feature_selection_graph":
            os.path.basename(
                FEATURE_SELECTION_GRAPH_PATH
            ),

        "fitness_graph":
            os.path.basename(
                FITNESS_GRAPH_PATH
            ),

        "confusion_matrix":
            os.path.basename(
                CONFUSION_MATRIX_PATH
            ),

        "distribution_graph":
            os.path.basename(
                DISTRIBUTION_GRAPH_PATH
            )
    }
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_data,
        file,
        indent=4,
        ensure_ascii=False
    )


print(
    "\nPSO JSON saved:"
)

print(
    JSON_PATH
)


# =============================================================================
# 60. VERIFY PSO PKL
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING PSO PKL")
print("=" * 80)


loaded_preprocessor = (
    joblib.load(
        PKL_PATH
    )
)


print(
    "\nPSO PKL loaded successfully."
)

print(
    "Optimization method:"
)

print(
    loaded_preprocessor[
        "method"
    ]
)

print(
    "\nPSO selected features:"
)

for feature in loaded_preprocessor[
    "selected_feature_columns"
]:

    print(
        " -",
        feature
    )


# =============================================================================
# 61. VERIFY PSO H5 MODEL
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING PSO H5 MODEL")
print("=" * 80)


loaded_model = (
    tf.keras.models.load_model(
        MODEL_PATH
    )
)


print(
    "\nPSO H5 model loaded successfully."
)

print(
    "Input shape:",
    loaded_model.input_shape
)

print(
    "Output shape:",
    loaded_model.output_shape
)


# =============================================================================
# 62. FINAL PSO SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("PSO PROJECT COMPLETED SUCCESSFULLY")
print("=" * 80)


print(
    "\nPARTICLE SWARM OPTIMIZATION"
)

print(
    "-" * 50
)

print(
    f"Swarm Size             : {PSO_SWARM_SIZE}"
)

print(
    f"Iterations             : {PSO_ITERATIONS}"
)

print(
    f"Inertia Range          : "
    f"{PSO_INERTIA_MAX} -> {PSO_INERTIA_MIN}"
)

print(
    f"Cognitive Coefficient : {PSO_COGNITIVE}"
)

print(
    f"Social Coefficient    : {PSO_SOCIAL}"
)

print(
    f"Best Fitness          : {best_fitness:.4f}"
)


print(
    "\nFEATURE SELECTION"
)

print(
    "-" * 50
)

print(
    "Features before PSO:",
    len(
        initial_feature_columns
    )
)

print(
    "Features after PSO:",
    len(
        selected_feature_columns
    )
)

print(
    "\nSelected Features:"
)

for feature in selected_feature_columns:

    print(
        " ✓",
        feature
    )


print(
    "\nPSO NEURAL NETWORK PERFORMANCE"
)

print(
    "-" * 50
)

print(
    f"Accuracy  : {nn_accuracy * 100:.2f}%"
)

print(
    f"Precision : {nn_precision * 100:.2f}%"
)

print(
    f"Recall    : {nn_recall * 100:.2f}%"
)

print(
    f"F1 Score  : {nn_f1 * 100:.2f}%"
)


if silhouette is not None:

    print(
        f"Silhouette: {silhouette:.4f}"
    )


print(
    "\nMODEL COMPARISON"
)

print(
    "-" * 80
)

print(
    comparison_df.to_string(
        index=False
    )
)


# =============================================================================
# 63. VERIFY ALL GENERATED PSO FILES
# =============================================================================

generated_files = [

    MODEL_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    RESULT_CSV_PATH,

    PREDICTION_CSV_PATH,

    COMPARISON_CSV_PATH,

    FEATURE_SELECTION_CSV_PATH,

    FITNESS_HISTORY_CSV_PATH,

    ACCURACY_GRAPH_PATH,

    LOSS_GRAPH_PATH,

    COMPARISON_GRAPH_PATH,

    HEATMAP_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_GRAPH_PATH,

    FEATURE_SELECTION_GRAPH_PATH,

    FITNESS_GRAPH_PATH,

    DISTRIBUTION_GRAPH_PATH,

    CONFUSION_MATRIX_PATH
]


print("\n" + "=" * 80)
print("GENERATED PSO FILES")
print("=" * 80)


for file_path in generated_files:

    if os.path.exists(
        file_path
    ):

        print(
            "✓",
            os.path.basename(
                file_path
            )
        )

    else:

        print(
            "✗",
            os.path.basename(
                file_path
            )
        )


print(
    "\nAll PSO outputs saved inside:"
)

print(
    PROJECT_DIR
)


print("\n" + "=" * 80)
print("DONE")
print("=" * 80)
```
