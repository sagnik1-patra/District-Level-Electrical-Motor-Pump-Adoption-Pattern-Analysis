```python
import os
import json
import warnings
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import joblib
import yaml

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    silhouette_score
)

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical


# ============================================================================
# 1. SETTINGS
# ============================================================================

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)


# ============================================================================
# 2. FILE PATHS
# ============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\District-Level Electrical Motor Pump Adoption Pattern Analysis"
)

DATASET_PATH = os.path.join(
    PROJECT_DIR,
    "5.49.csv"
)

MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "electrical_motor_pump_model.h5"
)

PKL_PATH = os.path.join(
    PROJECT_DIR,
    "electrical_motor_pump_preprocessor.pkl"
)

YAML_PATH = os.path.join(
    PROJECT_DIR,
    "electrical_motor_pump_config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR,
    "electrical_motor_pump_metadata.json"
)

PREDICTION_PATH = os.path.join(
    PROJECT_DIR,
    "electrical_motor_pump_predictions.csv"
)

ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "accuracy_graph.png"
)

LOSS_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "loss_graph.png"
)

CLUSTER_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "cluster_graph.png"
)

os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# ============================================================================
# 3. LOAD DATASET
# ============================================================================

print("=" * 80)
print("DISTRICT-LEVEL ELECTRICAL MOTOR PUMP ADOPTION PATTERN ANALYSIS")
print("=" * 80)

print("\nLoading dataset...")
print("Dataset path:")
print(DATASET_PATH)

df = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully.")
print("Dataset shape:", df.shape)

print("\nColumns:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")


# ============================================================================
# 4. BASIC DATASET INFORMATION
# ============================================================================

print("\n" + "=" * 80)
print("DATASET INFORMATION")
print("=" * 80)

print("\nFirst 5 rows:")
print(df.head())

print("\nDataset information:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nStatistical summary:")
print(df.describe(include="all").T)


# ============================================================================
# 5. CLEAN COLUMN NAMES
# ============================================================================

original_columns = list(df.columns)

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

print("\nCleaned column names:")

for col in df.columns:
    print(" -", col)


# ============================================================================
# 6. REMOVE COMPLETELY EMPTY COLUMNS
# ============================================================================

empty_columns = [
    col
    for col in df.columns
    if df[col].isna().all()
]

if empty_columns:
    print("\nRemoving completely empty columns:")
    print(empty_columns)

    df = df.drop(
        columns=empty_columns
    )


# ============================================================================
# 7. REMOVE DUPLICATES
# ============================================================================

initial_rows = len(df)

df = df.drop_duplicates().reset_index(drop=True)

removed_duplicates = initial_rows - len(df)

print(
    "\nDuplicate rows removed:",
    removed_duplicates
)


# ============================================================================
# 8. IDENTIFY NUMERIC COLUMNS
# ============================================================================

# Try converting object columns that actually contain numbers.

for col in df.columns:

    if df[col].dtype == "object":

        cleaned = (
            df[col]
            .astype(str)
            .str.strip()
            .str.replace(",", "", regex=False)
            .str.replace("%", "", regex=False)
        )

        converted = pd.to_numeric(
            cleaned,
            errors="coerce"
        )

        # Convert to numeric only if most non-null values can be parsed.
        valid_original = df[col].notna().sum()
        valid_numeric = converted.notna().sum()

        if (
            valid_original > 0
            and valid_numeric / valid_original >= 0.80
        ):
            df[col] = converted


numeric_columns = list(
    df.select_dtypes(
        include=np.number
    ).columns
)

categorical_columns = list(
    df.select_dtypes(
        exclude=np.number
    ).columns
)

print("\nNumeric columns:")
print(numeric_columns)

print("\nCategorical columns:")
print(categorical_columns)


# ============================================================================
# 9. HANDLE MISSING VALUES
# ============================================================================

for col in numeric_columns:

    median_value = df[col].median()

    if pd.isna(median_value):
        median_value = 0

    df[col] = df[col].fillna(
        median_value
    )


for col in categorical_columns:

    mode = df[col].mode()

    if len(mode) > 0:
        fill_value = mode.iloc[0]
    else:
        fill_value = "Unknown"

    df[col] = df[col].fillna(
        fill_value
    )


print(
    "\nTotal missing values after preprocessing:",
    int(df.isnull().sum().sum())
)


# ============================================================================
# 10. FIND MOTOR PUMP USAGE COLUMN
# ============================================================================

possible_target_columns = []

for col in df.columns:

    name = col.lower()

    if (
        "motor" in name
        and "pump" in name
    ):
        possible_target_columns.append(col)


print(
    "\nPossible electrical motor pump columns:"
)

for col in possible_target_columns:
    print(" -", col)


# Prefer a percentage column.

target_column = None

for col in possible_target_columns:

    if "%" in col or "percent" in col.lower():
        target_column = col
        break


if target_column is None and possible_target_columns:
    target_column = possible_target_columns[-1]


if target_column is None:

    raise ValueError(
        "\nCould not automatically identify the "
        "Electrical Motor Pump Use (%) column.\n"
        "Available columns are:\n"
        + "\n".join(df.columns)
    )


print(
    "\nSelected adoption indicator:",
    target_column
)


# ============================================================================
# 11. ENSURE TARGET IS NUMERIC
# ============================================================================

df[target_column] = pd.to_numeric(
    df[target_column],
    errors="coerce"
)

df[target_column] = df[target_column].fillna(
    df[target_column].median()
)

if df[target_column].isna().any():
    df[target_column] = df[target_column].fillna(0)


print("\nTarget statistics:")
print(df[target_column].describe())


# ============================================================================
# 12. FEATURE SELECTION
# ============================================================================

# We use the remaining useful numeric columns as input features.
# The target percentage itself is excluded from X.

feature_columns = [
    col
    for col in numeric_columns
    if col != target_column
]


# Remove constant numeric features.

usable_feature_columns = []

for col in feature_columns:

    if df[col].nunique() > 1:
        usable_feature_columns.append(col)


feature_columns = usable_feature_columns


# If the dataset does not have enough independent numeric features,
# we can use the adoption percentage for unsupervised pattern discovery
# and then train the classifier on available numerical characteristics.

if len(feature_columns) == 0:

    print(
        "\nWARNING: No independent varying numeric feature "
        "was detected."
    )

    # Create rank-based statistical features from the adoption value.
    # These are mainly for pattern-analysis demonstration on this small
    # academic dataset.

    df["Adoption_Rank"] = (
        df[target_column]
        .rank(method="average")
    )

    df["Adoption_Deviation"] = (
        df[target_column]
        - df[target_column].mean()
    )

    feature_columns = [
        "Adoption_Rank",
        "Adoption_Deviation"
    ]


print("\nFeatures used:")

for col in feature_columns:
    print(" -", col)


# ============================================================================
# 13. CREATE ADOPTION PATTERN LABELS USING K-MEANS
# ============================================================================

print("\n" + "=" * 80)
print("ADOPTION PATTERN DISCOVERY")
print("=" * 80)


adoption_values = df[
    [target_column]
].values


adoption_scaler = StandardScaler()

adoption_scaled = adoption_scaler.fit_transform(
    adoption_values
)


# Because this is a very small dataset, automatically select between
# 2 and 3 clusters depending on unique values.

unique_adoption_values = df[
    target_column
].nunique()


if unique_adoption_values >= 3:
    requested_clusters = 3
elif unique_adoption_values >= 2:
    requested_clusters = 2
else:
    requested_clusters = 1


if requested_clusters == 1:

    df["Cluster"] = 0

    print(
        "\nOnly one unique adoption value exists. "
        "All records belong to one pattern."
    )

else:

    kmeans = KMeans(
        n_clusters=requested_clusters,
        random_state=RANDOM_STATE,
        n_init=20
    )

    df["Cluster"] = kmeans.fit_predict(
        adoption_scaled
    )


# ============================================================================
# 14. ORDER CLUSTERS FROM LOW TO HIGH ADOPTION
# ============================================================================

cluster_means = (
    df.groupby("Cluster")[target_column]
    .mean()
    .sort_values()
)


ordered_clusters = list(
    cluster_means.index
)


cluster_name_mapping = {}


if len(ordered_clusters) == 1:

    cluster_name_mapping[
        ordered_clusters[0]
    ] = "Uniform Adoption"

elif len(ordered_clusters) == 2:

    cluster_name_mapping[
        ordered_clusters[0]
    ] = "Low Adoption"

    cluster_name_mapping[
        ordered_clusters[1]
    ] = "High Adoption"

else:

    cluster_name_mapping[
        ordered_clusters[0]
    ] = "Low Adoption"

    cluster_name_mapping[
        ordered_clusters[1]
    ] = "Moderate Adoption"

    cluster_name_mapping[
        ordered_clusters[2]
    ] = "High Adoption"


df["Adoption_Level"] = df[
    "Cluster"
].map(cluster_name_mapping)


print("\nCluster mean adoption values:")

for cluster_id, mean_value in cluster_means.items():

    print(
        f"Cluster {cluster_id}: "
        f"{mean_value:.4f} "
        f"-> {cluster_name_mapping[cluster_id]}"
    )


print("\nAdoption level distribution:")

print(
    df["Adoption_Level"]
    .value_counts()
)


# ============================================================================
# 15. SILHOUETTE SCORE
# ============================================================================

silhouette = None

if (
    requested_clusters > 1
    and len(np.unique(df["Cluster"])) > 1
    and len(df) > requested_clusters
):

    try:

        silhouette = silhouette_score(
            adoption_scaled,
            df["Cluster"]
        )

        print(
            "\nSilhouette Score:",
            round(silhouette, 4)
        )

    except Exception as e:

        print(
            "\nSilhouette score could not be calculated:",
            e
        )


# ============================================================================
# 16. ENCODE TARGET LABEL
# ============================================================================

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    df["Adoption_Level"]
)


class_names = list(
    label_encoder.classes_
)

num_classes = len(class_names)


print("\nEncoded classes:")

for i, name in enumerate(class_names):
    print(i, "->", name)


# ============================================================================
# 17. PREPARE FEATURE MATRIX
# ============================================================================

X = df[
    feature_columns
].copy()


feature_scaler = StandardScaler()

X_scaled = feature_scaler.fit_transform(
    X
)


print(
    "\nFeature matrix shape:",
    X_scaled.shape
)


# ============================================================================
# 18. TRAIN / TEST SPLIT
# ============================================================================

print("\n" + "=" * 80)
print("TRAIN / TEST PREPARATION")
print("=" * 80)


class_counts = pd.Series(y).value_counts()

print("\nClass counts:")
print(class_counts)


# Stratification is only possible if every class has at least 2 samples.

can_stratify = (
    len(class_counts) > 1
    and class_counts.min() >= 2
)


if len(df) >= 10:

    test_size = 0.25

else:

    test_size = 0.30


if num_classes > 1:

    X_train, X_test, y_train, y_test = train_test_split(
        X_scaled,
        y,
        test_size=test_size,
        random_state=RANDOM_STATE,
        stratify=y if can_stratify else None
    )

else:

    # Neural-network classification is not meaningful with only one class.
    # Still preserve the entire dataset and artifacts.

    X_train = X_scaled
    X_test = X_scaled

    y_train = y
    y_test = y


print(
    "\nTraining samples:",
    len(X_train)
)

print(
    "Testing samples:",
    len(X_test)
)


# ============================================================================
# 19. ONE-HOT ENCODE LABELS
# ============================================================================

if num_classes > 1:

    y_train_cat = to_categorical(
        y_train,
        num_classes=num_classes
    )

    y_test_cat = to_categorical(
        y_test,
        num_classes=num_classes
    )


# ============================================================================
# 20. BUILD NEURAL NETWORK
# ============================================================================

print("\n" + "=" * 80)
print("BUILDING NEURAL NETWORK")
print("=" * 80)


if num_classes > 1:

    model = Sequential(
        [
            Input(
                shape=(X_train.shape[1],)
            ),

            Dense(
                32,
                activation="relu"
            ),

            Dropout(
                0.20
            ),

            Dense(
                16,
                activation="relu"
            ),

            Dropout(
                0.10
            ),

            Dense(
                num_classes,
                activation="softmax"
            )
        ]
    )


    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.001
        ),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )


    model.summary()


# ============================================================================
# 21. TRAIN MODEL
# ============================================================================

history = None


if num_classes > 1:

    print("\nTraining model...")


    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=25,
        restore_best_weights=True,
        verbose=1
    )


    history = model.fit(
        X_train,
        y_train_cat,
        validation_split=0.20,
        epochs=200,
        batch_size=8,
        callbacks=[
            early_stopping
        ],
        verbose=1
    )


# ============================================================================
# 22. MODEL EVALUATION
# ============================================================================

accuracy = None
predicted_classes = None


if num_classes > 1:

    print("\n" + "=" * 80)
    print("MODEL EVALUATION")
    print("=" * 80)


    probabilities = model.predict(
        X_test,
        verbose=0
    )


    predicted_classes = np.argmax(
        probabilities,
        axis=1
    )


    accuracy = accuracy_score(
        y_test,
        predicted_classes
    )


    print(
        "\nTest Accuracy:",
        round(accuracy * 100, 2),
        "%"
    )


    print("\nClassification Report:")

    print(
        classification_report(
            y_test,
            predicted_classes,
            labels=np.arange(num_classes),
            target_names=class_names,
            zero_division=0
        )
    )


    print("\nConfusion Matrix:")

    print(
        confusion_matrix(
            y_test,
            predicted_classes,
            labels=np.arange(num_classes)
        )
    )


# ============================================================================
# 23. TRAINING ACCURACY GRAPH
# ============================================================================

if history is not None:

    plt.figure(
        figsize=(10, 6)
    )

    plt.plot(
        history.history["accuracy"],
        label="Training Accuracy",
        linewidth=2
    )

    plt.plot(
        history.history["val_accuracy"],
        label="Validation Accuracy",
        linewidth=2
    )

    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Accuracy"
    )

    plt.title(
        "Electrical Motor Pump Adoption Classification Accuracy"
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        ACCURACY_GRAPH_PATH,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================================
# 24. LOSS GRAPH
# ============================================================================

if history is not None:

    plt.figure(
        figsize=(10, 6)
    )

    plt.plot(
        history.history["loss"],
        label="Training Loss",
        linewidth=2
    )

    plt.plot(
        history.history["val_loss"],
        label="Validation Loss",
        linewidth=2
    )

    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Loss"
    )

    plt.title(
        "Electrical Motor Pump Adoption Classification Loss"
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    plt.savefig(
        LOSS_GRAPH_PATH,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================================
# 25. ADOPTION CLUSTER GRAPH
# ============================================================================

plt.figure(
    figsize=(11, 6)
)


sorted_df = df.sort_values(
    target_column
).reset_index(drop=True)


plt.scatter(
    np.arange(len(sorted_df)),
    sorted_df[target_column],
    c=sorted_df["Cluster"],
    s=80
)


plt.xlabel(
    "District Record"
)

plt.ylabel(
    target_column
)

plt.title(
    "District-Level Electrical Motor Pump Adoption Patterns"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    CLUSTER_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================================
# 26. GENERATE PREDICTIONS FOR COMPLETE DATASET
# ============================================================================

if num_classes > 1:

    full_probabilities = model.predict(
        X_scaled,
        verbose=0
    )

    full_predictions = np.argmax(
        full_probabilities,
        axis=1
    )

    predicted_labels = label_encoder.inverse_transform(
        full_predictions
    )

else:

    full_probabilities = np.ones(
        (len(df), 1)
    )

    predicted_labels = np.array(
        [class_names[0]] * len(df)
    )


prediction_df = df.copy()

prediction_df[
    "Predicted_Adoption_Level"
] = predicted_labels


if num_classes > 1:

    prediction_df[
        "Prediction_Confidence"
    ] = np.max(
        full_probabilities,
        axis=1
    )

else:

    prediction_df[
        "Prediction_Confidence"
    ] = 1.0


prediction_df.to_csv(
    PREDICTION_PATH,
    index=False
)


print(
    "\nPrediction file saved:"
)

print(
    PREDICTION_PATH
)


# ============================================================================
# 27. SAVE H5 MODEL
# ============================================================================

print("\n" + "=" * 80)
print("SAVING MODEL AND ARTIFACTS")
print("=" * 80)


if num_classes > 1:

    model.save(
        MODEL_PATH
    )

else:

    # If the dataset unexpectedly contains only one class, create a tiny
    # single-output Keras model so that the required H5 artifact is still
    # generated.

    model = Sequential(
        [
            Input(
                shape=(X_scaled.shape[1],)
            ),

            Dense(
                8,
                activation="relu"
            ),

            Dense(
                1,
                activation="sigmoid"
            )
        ]
    )

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    dummy_y = np.ones(
        len(X_scaled)
    )

    model.fit(
        X_scaled,
        dummy_y,
        epochs=10,
        batch_size=8,
        verbose=0
    )

    model.save(
        MODEL_PATH
    )


print(
    "\nH5 model saved:"
)

print(
    MODEL_PATH
)


# ============================================================================
# 28. SAVE PKL PREPROCESSOR
# ============================================================================

preprocessor = {

    "feature_scaler":
        feature_scaler,

    "adoption_scaler":
        adoption_scaler,

    "label_encoder":
        label_encoder,

    "feature_columns":
        feature_columns,

    "target_column":
        target_column,

    "cluster_name_mapping":
        cluster_name_mapping,

    "original_columns":
        original_columns,

    "processed_columns":
        list(df.columns),

    "random_state":
        RANDOM_STATE
}


# Save KMeans only if it exists.

if requested_clusters > 1:

    preprocessor[
        "kmeans"
    ] = kmeans


joblib.dump(
    preprocessor,
    PKL_PATH
)


print(
    "\nPKL preprocessor saved:"
)

print(
    PKL_PATH
)


# ============================================================================
# 29. SAVE YAML CONFIGURATION
# ============================================================================

yaml_config = {

    "project": {
        "name":
            "District-Level Electrical Motor Pump Adoption Pattern Analysis",

        "type":
            "Machine Learning and Deep Learning Pattern Analysis",

        "random_state":
            int(RANDOM_STATE)
    },

    "dataset": {
        "path":
            DATASET_PATH,

        "rows":
            int(df.shape[0]),

        "columns":
            int(df.shape[1]),

        "target_column":
            target_column,

        "feature_columns":
            feature_columns
    },

    "clustering": {
        "algorithm":
            "KMeans",

        "number_of_clusters":
            int(requested_clusters),

        "cluster_labels":
            {
                str(k): str(v)
                for k, v
                in cluster_name_mapping.items()
            },

        "silhouette_score":
            (
                float(silhouette)
                if silhouette is not None
                else None
            )
    },

    "model": {
        "framework":
            "TensorFlow / Keras",

        "model_type":
            "Feedforward Neural Network",

        "input_features":
            int(len(feature_columns)),

        "number_of_classes":
            int(num_classes),

        "classes":
            [
                str(x)
                for x in class_names
            ],

        "optimizer":
            "Adam",

        "loss":
            (
                "categorical_crossentropy"
                if num_classes > 1
                else "binary_crossentropy"
            ),

        "epochs":
            200,

        "batch_size":
            8
    },

    "output_files": {
        "model":
            MODEL_PATH,

        "preprocessor":
            PKL_PATH,

        "metadata":
            JSON_PATH,

        "predictions":
            PREDICTION_PATH,

        "accuracy_graph":
            ACCURACY_GRAPH_PATH,

        "loss_graph":
            LOSS_GRAPH_PATH,

        "cluster_graph":
            CLUSTER_GRAPH_PATH
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_config,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print(
    "\nYAML configuration saved:"
)

print(
    YAML_PATH
)


# ============================================================================
# 30. SAVE JSON METADATA
# ============================================================================

metadata = {

    "project_name":
        "District-Level Electrical Motor Pump Adoption Pattern Analysis",

    "dataset_name":
        os.path.basename(
            DATASET_PATH
        ),

    "dataset_path":
        DATASET_PATH,

    "dataset_rows":
        int(df.shape[0]),

    "dataset_columns":
        int(df.shape[1]),

    "target_column":
        target_column,

    "feature_columns":
        [
            str(x)
            for x in feature_columns
        ],

    "number_of_features":
        int(len(feature_columns)),

    "number_of_classes":
        int(num_classes),

    "classes":
        [
            str(x)
            for x in class_names
        ],

    "cluster_count":
        int(requested_clusters),

    "cluster_mapping":
        {
            str(k): str(v)
            for k, v
            in cluster_name_mapping.items()
        },

    "cluster_mean_adoption":
        {
            str(k): float(v)
            for k, v
            in cluster_means.items()
        },

    "silhouette_score":
        (
            float(silhouette)
            if silhouette is not None
            else None
        ),

    "test_accuracy":
        (
            float(accuracy)
            if accuracy is not None
            else None
        ),

    "model": {
        "type":
            "Artificial Neural Network",

        "framework":
            "TensorFlow Keras",

        "format":
            "HDF5 (.h5)"
    },

    "preprocessing": {
        "numeric_missing_values":
            "Median imputation",

        "categorical_missing_values":
            "Mode imputation",

        "feature_scaling":
            "StandardScaler",

        "label_encoding":
            "LabelEncoder"
    },

    "artifacts": {
        "h5":
            os.path.basename(
                MODEL_PATH
            ),

        "pkl":
            os.path.basename(
                PKL_PATH
            ),

        "yaml":
            os.path.basename(
                YAML_PATH
            ),

        "json":
            os.path.basename(
                JSON_PATH
            ),

        "predictions":
            os.path.basename(
                PREDICTION_PATH
            )
    }
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )


print(
    "\nJSON metadata saved:"
)

print(
    JSON_PATH
)


# ============================================================================
# 31. VERIFY SAVED PKL
# ============================================================================

print("\n" + "=" * 80)
print("VERIFYING SAVED PREPROCESSOR")
print("=" * 80)


loaded_preprocessor = joblib.load(
    PKL_PATH
)


print(
    "\nPKL loaded successfully."
)

print(
    "Saved feature columns:"
)

for feature in loaded_preprocessor[
    "feature_columns"
]:
    print(" -", feature)


# ============================================================================
# 32. VERIFY SAVED H5 MODEL
# ============================================================================

print("\n" + "=" * 80)
print("VERIFYING SAVED H5 MODEL")
print("=" * 80)


loaded_model = tf.keras.models.load_model(
    MODEL_PATH
)


print(
    "\nH5 model loaded successfully."
)

print(
    "Input shape:",
    loaded_model.input_shape
)

print(
    "Output shape:",
    loaded_model.output_shape
)


# ============================================================================
# 33. FINAL SUMMARY
# ============================================================================

print("\n" + "=" * 80)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 80)


print(
    "\nDataset:",
    DATASET_PATH
)

print(
    "\nRows:",
    len(df)
)

print(
    "Target:",
    target_column
)

print(
    "Number of features:",
    len(feature_columns)
)

print(
    "Number of adoption classes:",
    num_classes
)


if accuracy is not None:

    print(
        "Test Accuracy:",
        f"{accuracy * 100:.2f}%"
    )


if silhouette is not None:

    print(
        "Silhouette Score:",
        f"{silhouette:.4f}"
    )


print("\nGenerated files:")

generated_files = [
    MODEL_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH,
    PREDICTION_PATH,
    ACCURACY_GRAPH_PATH,
    LOSS_GRAPH_PATH,
    CLUSTER_GRAPH_PATH
]


for path in generated_files:

    if os.path.exists(path):

        print(
            "✓",
            os.path.basename(path)
        )

    else:

        print(
            "✗",
            os.path.basename(path)
        )


print("\nAll outputs are stored in:")

print(
    PROJECT_DIR
)

print("\n" + "=" * 80)
```
